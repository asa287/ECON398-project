Now, we want to get the working status in 2020 for each individual

```r
df$agegrp <- as_factor(df$agegrp)  # This retains Stata labels as factor levels
```

```r
df$agegrp
```

```r
df$cow
```

```r
df <- df %>% #<-- pipeline, telling R I want to generate a new column 
    mutate(work = ifelse(cow != 9, 1, 0)) # if the condition in the first argument is true, the variable will be 1, if it not true, it will be 0. 
```

```r
df$work %>% table()
# same as table(df$work)
```

Let's see how ages are distributed

```r
# Create histogram with proper age labels
ggplot(df, aes(x = agegrp)) +
    geom_bar() +
    labs(title = "Age Group Distribution", x = "Age Group", y = "Count") +
    theme_minimal()+ #aesthetic
    theme(axis.text.x = element_text(angle = 45, hjust = 1)) #aesthetic
```

```r
### Labor force participation by age group

# Calculate the share of working individuals for each age group
work_share <- df %>%
    filter(agegrp != "Not available") %>% 
    group_by(agegrp) %>%
    summarise(share_working = mean(work, na.rm = TRUE))  # Compute share of working individuals

work_share
```

```r
### Labor force participation by age group

# Calculate the share of working individuals for each age group
work_share <- df %>%
    filter(agegrp != "Not available") %>% 
    #group_by(agegrp) %>%
    summarise(share_working = mean(work, na.rm = TRUE))  # Compute share of working individuals

work_share
```

```r
# Plot the share of working individuals by age group
ggplot(work_share, aes(x = agegrp, y = share_working)) +
  geom_col() +
  labs(title = "Share of Working Individuals by Age Group",
       x = "Age Group", y = "Share Working") +
    theme_minimal()+ #aesthetic
  theme(axis.text.x = element_text(angle = 45, hjust = 1))  #aesthetic
```

How about by gender?

```r
df$Gender <- as_factor(df$Gender)
```

```r
# Calculate the share of working individuals for each age group and gender
work_share <- df %>%
  filter(agegrp != "Not available") %>%
  group_by(agegrp, Gender) %>%
  summarise(share_working = mean(work, na.rm = TRUE), .groups = "drop")  # Compute share of working individuals

work_share
```

```r
# Plot the share of working individuals by age group, with separate panels for gender
ggplot(work_share) +
  geom_col(position = "dodge", aes(x = agegrp, y = share_working, fill = Gender)) +
  geom_vline(xintercept = 16.5, color = "red", linetype = "dashed", size = 1) +  # Add red line
  facet_wrap(~Gender, nrow = 2) +  # Creates side-by-side panels for gender
  labs(title = "Share of Working Individuals by Age Group and Gender",
       x = "Age Group", y = "Share Working") +
    theme_minimal()+ #aesthetic
  theme(axis.text.x = element_text(angle = 45, hjust = 1), #aesthetic
       legend.position = "bottom") #aesthetic
```
